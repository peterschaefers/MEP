```python
    def process_CMAES_data(dataset, population_size, qubit_names=None):
        """
        Processes fidelity data for n qubits based on the given dataset and population size.

        Parameters:
            dataset: An object containing data attributes (e.g., dataset.x0.data, dataset.y0.data, etc.).
            population_size: The size of the population.
            nr_qubits: The number of qubits.

        Returns:
            A dictionary containing mean and population fidelity data for all qubit_names and shared parameters.
        """

        if qubit_names is None:
            if hasattr(dataset, "elements"):
                qubit_names = dataset.elements
            else:
                raise ValueError("No qubit names provided and dataset has no 'elements' attribute.")

        # Extract parameters shared across all qubits
        shared_data = {
            'objective': dataset.y0.data,
        }

        # Initialize dictionary to hold data for each qubit
        fidelity_data = {'shared': shared_data}

        name_to_var = {dataset[var].attrs.get("name"): var for var in dataset.data_vars}
        name_to_coord = {dataset[coord].attrs.get("name"): coord for coord in dataset.coords}

        for qubit in qubit_names:
            # Extract fidelity data for the current qubit
            fidelity_data[qubit] = {
                'freq': dataset[name_to_coord.get(f"frequency_{qubit}")].values,
                'ampl': dataset[name_to_coord.get(f"amplitude_{qubit}")].values,
                'dur': dataset[name_to_coord.get(f"duration_{qubit}")].values,
                'assignfid_av': dataset[name_to_var.get(f"av_assignment_{qubit}")].values,
                'qndpifid': dataset[name_to_var.get(f"qnd_pi_{qubit}")].values,
            }

            # Extract data of the mean
            fidelity_data[qubit]['mean'] = {
                key: fidelity_data[qubit][key][population_size::population_size + 1]
                for key in
                ['freq', 'ampl', 'dur', 'assignfid_av', 'qndpifid']
            }

            # Extract data of the population
            fidelity_data[qubit]['pop'] = {
                key: np.array([
                    fidelity_data[qubit][key][i::population_size + 1]
                    for i in range(population_size)
                ])
                for key in
                ['freq', 'ampl', 'dur', 'assignfid_av', 'qndpifid']
            }

        return fidelity_data

```

```python
class ReadoutOptimizationAnalysis(BaseAnalysis):
    """
    Behaviour class, contains analysis methods for ssro-qnd-pi experiment.
    """

    # region Class Constructor
    def __init__(
            self,
            population_size: int,
            nr_qubits: int,
            nr_generations: int,
            qubit_names: Optional[List[str]] = None,
            plot_population: Optional[bool] = False,
            *args,
            **kwargs,
    ):
        super().__init__(*args, **kwargs)
        self.population_size: int = population_size
        self.nr_qubits: int = nr_qubits
        self.nr_generations: int = nr_generations
        self.qubit_names: List[str] = qubit_names
        self.plot_population: bool = plot_population

    # endregion

    # region Interface Methods
    def process_CMAES_data(self, population_size):
        """
        Processes fidelity data for n qubits based on the given dataset and population size.

        Parameters:
            dataset: An object containing data attributes (e.g., dataset.x0.data, dataset.y0.data, etc.).
            population_size: The size of the population.
            nr_qubits: The number of qubits.

        Returns:
            A dictionary containing mean and population fidelity data for all qubit_names and shared parameters.
        """
        qubit_names = self.qubit_names if self.qubit_names else self.dataset.elements

        # Extract parameters shared across all qubits
        shared_data = {
            'objective': self.dataset.y0.data,
        }

        # Initialize dictionary to hold data for each qubit
        fidelity_data = {'shared': shared_data}

        name_to_var = {self.dataset[var].attrs.get("name"): var for var in self.dataset.data_vars}
        name_to_coord = {self.dataset[coord].attrs.get("name"): coord for coord in self.dataset.coords}

        print(qubit_names)
        for qubit in qubit_names:
            # Extract fidelity data for the current qubit
            fidelity_data[qubit] = {
                'freq': self.dataset[name_to_coord.get(f"frequency_{qubit}")].values,
                'ampl': self.dataset[name_to_coord.get(f"amplitude_{qubit}")].values,
                'dur': self.dataset[name_to_coord.get(f"duration_{qubit}")].values,
                'assignfid_av': self.dataset[name_to_var.get(f"av_assignment_{qubit}")].values,
                'assignfid_01': self.dataset[name_to_var.get(f"01_assignment_{qubit}")].values,
                'assignfid_02': self.dataset[name_to_var.get(f"02_assignment_{qubit}")].values,
                'assignfid_12': self.dataset[name_to_var.get(f"12_assignment_{qubit}")].values,
                'qndpifid': self.dataset[name_to_var.get(f"qnd_pi_{qubit}")].values,
            }

            # Extract data of the mean
            fidelity_data[qubit]['mean'] = {
                key: fidelity_data[qubit][key][population_size::population_size + 1]
                for key in
                ['freq', 'ampl', 'dur', 'assignfid_av', 'assignfid_01', 'assignfid_02', 'assignfid_12', 'qndpifid']
            }

            # Extract data of the population
            fidelity_data[qubit]['pop'] = {
                key: np.array([
                    fidelity_data[qubit][key][i::population_size + 1]
                    for i in range(population_size)
                ])
                for key in
                ['freq', 'ampl', 'dur', 'assignfid_av', 'assignfid_01', 'assignfid_02', 'assignfid_12', 'qndpifid']
            }

        return fidelity_data

    def set_plotstyle(self):
        fig_size_dim = 8
        golden_ratio = (1 + np.sqrt(5)) / 2
        fig_size = (fig_size_dim, fig_size_dim / golden_ratio)

        font_size = 8
        dpi = 500

        params = {'figure.figsize': fig_size,
                  'figure.dpi': dpi,
                  'savefig.dpi': dpi,
                  'font.size': font_size,
                  'font.family': "sans-serif",
                  'figure.titlesize': font_size,
                  'legend.fontsize': font_size,
                  'axes.labelsize': font_size,
                  'axes.titlesize': font_size,
                  'xtick.labelsize': font_size,
                  'ytick.labelsize': font_size,
                  }
        return params

    def run_fitting(self):
        """
        Fit the data to :class:`~quantify_core.analysis.fitting_models.CosineModel`,
        and to :class:`~lmfit.models.LinearModel`, then chooses the best fit.
        """
        processed_data = self.process_CMAES_data(self.population_size)
        return processed_data


    def analyze_fit_results(self):
        """
        Checks fit success and populates :code:`.quantities_of_interest`.
        """
        pass

    def create_figures(self):
        """
        Create a figure showing the cosine and linear fit.
        """
        plot_params = self.set_plotstyle()
        processed_data = self.run_fitting()
        qubit_names = self.qubit_names if self.qubit_names else self.dataset.elements

        # --- Make flat-lightness YlGnBu (darker) ---
        orig_cmap = mpl.colormaps['YlGnBu']
        N = 256
        rgb = orig_cmap(np.linspace(0, 1, N))[:, :3]

        fixed_lightness = 0.4  # darker; try 0.5 for medium
        new_rgb = []
        for r, g, b in rgb:
            h, l, s = colorsys.rgb_to_hls(r, g, b)
            new_rgb.append(colorsys.hls_to_rgb(h, fixed_lightness, s))

        flat_dark_cmap = ListedColormap(new_rgb, name='YlGnBu_flat_dark')

        spacing = np.linspace(0.1, 0.9, self.nr_qubits)
        colors = [flat_dark_cmap(s) for s in spacing]

        with plt.rc_context(plot_params):
            """
            Create figure for fidelities
            """

            fig, axes = plt.subplots(ncols=2, nrows=2, sharex=True)
            ax1 = axes[0][0]
            ax2 = axes[0][1]
            ax3 = axes[1][0]
            ax4 = axes[1][1]

            fig.set_size_inches(6.2, 4.8)

            for idx, qubit in enumerate(qubit_names):
                color = colors[idx]

                ax1.plot(range(self.nr_generations), processed_data[qubit]['mean']['assignfid_av'][:self.nr_generations], lw=1, label=qubit,
                         color=color)
                ax2.plot(range(self.nr_generations), 1 - processed_data[qubit]['mean']['assignfid_av'][:self.nr_generations], lw=1, label=qubit,
                         color=color)

                if self.plot_population:
                    for i in range(self.population_size):
                        ax1.plot(range(self.nr_generations), processed_data[qubit]['pop']['assignfid_av'][i][:self.nr_generations],".", color=color, alpha=0.1)
                        ax2.plot(range(self.nr_generations), 1 - processed_data[qubit]['pop']['assignfid_av'][i][:self.nr_generations],".",color=color, alpha=0.1)

            # Axis formatting
            ax1.grid(which='both', alpha=.2)
            ax1.set_ylabel(r'$\bar{\mathrm{F}}_\mathrm{assign}$')
            ax1.set_ylim(0.5, 1)
            ax1.legend(fontsize="small", ncols=2)

            ax2.grid(which='both', alpha=.2)
            ax2.set_ylabel(r'$1 - \bar{\mathrm{F}}_\mathrm{assign}$')
            ax2.set_yscale('log')
            ax2.set_ylim(1e-4, 0.5)

            for idx, qubit in enumerate(qubit_names):
                color = colors[idx]

                ax3.plot(range(self.nr_generations), processed_data[qubit]['mean']['qndpifid'][:self.nr_generations], lw=1, label=qubit,
                         color=color)
                ax4.plot(range(self.nr_generations), 1 - processed_data[qubit]['mean']['qndpifid'][:self.nr_generations], lw=1, label=qubit,
                         color=color)

                if self.plot_population:
                    for i in range(self.population_size):
                        ax3.plot(range(self.nr_generations), processed_data[qubit]['pop']['qndpifid'][i][:self.nr_generations],".", color=color, alpha=0.1)
                        ax4.plot(range(self.nr_generations), 1 - processed_data[qubit]['pop']['qndpifid'][i][:self.nr_generations],".", color=color, alpha=0.1)

            # Axis formatting
            ax3.grid(which='both', alpha=.2)
            ax3.set_ylabel(r'$\mathrm{F}_\mathrm{QND-\pi}$')
            ax3.set_xlabel(r'N$_\mathrm{gen}$')
            ax3.set_ylim(0.5, 1)

            ax4.grid(which='both', alpha=.2)
            ax4.set_ylabel(r'1-$\mathrm{F}_\mathrm{QND-\pi}$')
            ax4.set_xlabel(r'N$_\mathrm{gen}$')
            ax4.set_yscale('log')
            ax4.set_ylim(1e-4, 0.5)

            fig.suptitle(f"tuid: {self.dataset.tuid}")
            fig.tight_layout()

            self.figs_mpl["Fidelity evolution"] = fig
            self.axs_mpl["Fidelity evolution"] = ax1

            """
            Create figure for readout parameters
            """
            fig2, axes2 = plt.subplots(ncols=3, nrows=1)
            ax1 = axes2[0]
            ax2 = axes2[1]
            ax3 = axes2[2]

            fig.set_size_inches(6.2, 2.4)

            for idx, qubit in enumerate(qubit_names):
                color = colors[idx]
                ax1.plot(range(self.nr_generations), (processed_data[qubit]['mean']['freq'][:self.nr_generations]-processed_data[qubit]['mean']['freq'][:self.nr_generations][-1]) * 1e-6, lw=1, label=qubit,
                         color=color)
                ax2.plot(range(self.nr_generations), processed_data[qubit]['mean']['ampl'][:self.nr_generations], lw=1, label=qubit,
                         color=color)
                ax3.plot(range(self.nr_generations), processed_data[qubit]['mean']['dur'][:self.nr_generations] * 1e9, lw=1, label=qubit,
                         color=color)

                if self.plot_population:
                    for i in range(self.population_size):
                        ax1.plot(range(self.nr_generations), (processed_data[qubit]['pop']['freq'][i][:self.nr_generations]-processed_data[qubit]['pop']['freq'][i][:self.nr_generations][-1]) * 1e-6,".", color=color, alpha=0.1)
                        ax2.plot(range(self.nr_generations), processed_data[qubit]['pop']['ampl'][i][:self.nr_generations], ".",color=color, alpha=0.1)
                        ax3.plot(range(self.nr_generations), processed_data[qubit]['pop']['dur'][i][:self.nr_generations] * 1e9,".", color=color, alpha=0.1)

            ax1.set_ylabel(r'$\Delta$f (MHz)')
            ax1.set_xlabel(r'N$_\mathrm{gen}$')
            ax1.grid(alpha=.2)

            ax2.set_ylabel('A (a.u.)')
            ax2.set_xlabel(r'N$_\mathrm{gen}$')
            ax2.grid(alpha=.2)

            ax3.set_ylabel('dt (ns)')
            ax3.set_xlabel(r'N$_\mathrm{gen}$')
            ax3.grid(alpha=.2)
            ax3.legend(fontsize="x-small", ncols=2)

            fig2.suptitle(f"tuid: {self.dataset.tuid}")
            fig2.tight_layout()

            self.figs_mpl["Readout parameters"] = fig2
            self.axs_mpl["Readout parameters"] = ax1

            """
            Create figures for infidelity
            """

            fidelities = ["assignfid_av", "assignfid_01", "assignfid_02", "assignfid_12", "qndpifid"]
            nr_gen = 10 #Number of generations to average over

            qubit_names = np.sort(self.dataset.elements)

            fidelity_dict = {}

            for fidelity in fidelities:
                tmp_mean = []
                for qubit in qubit_names:
                    mean_fidelity = np.mean(processed_data[qubit]['mean'][fidelity][-nr_gen:])
                    tmp_mean.append(mean_fidelity)

                fidelity_dict[fidelity] = tmp_mean

            fig3, ax = plt.subplots(ncols=1, nrows=1)

            ax.ecdf(1 - np.array(fidelity_dict[fidelities[0]]), color="royalblue",
                    label=r"$\bar{\text{F}}_{\text{assign}}$")
            ax.vlines(np.mean(1 - np.array(fidelity_dict[fidelities[0]])), 0, 1, ls="--", color="royalblue")

            ax.ecdf(1 - np.array(fidelity_dict[fidelities[1]]), color="dimgrey", label=r"F$_{\text{01}}$")
            ax.vlines(np.mean(1 - np.array(fidelity_dict[fidelities[1]])), 0, 1, ls="--", color="dimgrey")

            ax.ecdf(1 - np.array(fidelity_dict[fidelities[2]]), color="teal", label=r"F$_{\text{02}}$")
            ax.vlines(np.mean(1 - np.array(fidelity_dict[fidelities[2]])), 0, 1, ls="--", color="teal")

            ax.ecdf(1 - np.array(fidelity_dict[fidelities[3]]), color="purple", label=r"F$_{\text{12}}$")
            ax.vlines(np.mean(1 - np.array(fidelity_dict[fidelities[3]])), 0, 1, ls="--", color="purple")

            ax.ecdf(1 - np.array(fidelity_dict[fidelities[4]]), color="firebrick", label=r"F$_{\text{QND-\pi}}$")
            ax.vlines(np.mean(1 - np.array(fidelity_dict[fidelities[4]])), 0, 1, ls="--", color="firebrick")

            ax.set_xlabel(r"1-F")
            ax.legend(fontsize="x-small", ncols=2)
            ax.set_xscale('log')

            fig3.suptitle(f"tuid: {self.dataset.tuid}")
            fig3.tight_layout()

            self.figs_mpl["Infidelity"] = fig3
            self.axs_mpl["Infidelity"] = ax


    # region Static Class Methods
    @staticmethod
    def set_attribute_value(obj: object, attr_path: str, value: Any) -> None:
        """
        Sets the value of a potentially nested attribute.

        :param obj: The root object on which to set the attribute.
        :param attr_path: The dotted path to the attribute (e.g., "cz.snz_pulse_duration").
        :param value: The value to set for the attribute.
        :raises AttributeError: If any attribute in the path does not exist.
        """
        # Split the attribute path into components
        attrs = attr_path.split(".")

        # Traverse to the second-to-last attribute
        for attr in attrs[:-1]:
            obj = getattr(obj, attr)

        # Set the value to the last attribute in the path
        setattr(obj, attrs[-1], value)

    # endregion
```
